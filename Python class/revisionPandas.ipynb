```python
import pandas as pd
import os

FILE_NAME = "bank1.csv"


# -----------------------------
# Load Data
# -----------------------------
def load_data():
    if not os.path.exists(FILE_NAME) or os.path.getsize(FILE_NAME) == 0:
        df = pd.DataFrame(columns=["A/C #", "name", "password", "A/C bal"])
        df.to_csv(FILE_NAME, index=False)
        return {}

    df = pd.read_csv(FILE_NAME)

    bank = {}

    for _, row in df.iterrows():
        account = str(row["A/C #"])
        bank[account] = {
            "name": row["name"],
            "password": str(row["password"]),
            "bal": float(row["A/C bal"])
        }

    return bank


# -----------------------------
# Save Data
# -----------------------------
def save_data(bank):
    accounts = []
    names = []
    passwords = []
    balances = []

    for account, info in bank.items():
        accounts.append(account)
        names.append(info["name"])
        passwords.append(info["password"])
        balances.append(info["bal"])

    df = pd.DataFrame({
        "A/C #": accounts,
        "name": names,
        "password": passwords,
        "A/C bal": balances
    })

    df.to_csv(FILE_NAME, index=False)


# -----------------------------
# Create Account
# -----------------------------
def create_account(bank):
    account = input("Enter Account Number: ")

    if account in bank:
        print("❌ Account already exists.")
        return

    name = input("Enter Account Holder Name: ")
    password = input("Create Password: ")
    balance = float(input("Opening Balance: "))

    bank[account] = {
        "name": name,
        "password": password,
        "bal": balance
    }

    print("✅ Account Created Successfully.")


# -----------------------------
# Balance Inquiry
# -----------------------------
def balance_inquiry(bank, account):
    print("\n------ ACCOUNT DETAILS ------")
    print("Name    :", bank[account]["name"])
    print("Account :", account)
    print("Balance :", bank[account]["bal"])


# -----------------------------
# Deposit
# -----------------------------
def deposit(bank, account):
    amount = float(input("Enter Deposit Amount: "))

    if amount <= 0:
        print("Invalid amount.")
        return

    bank[account]["bal"] += amount

    print("Deposit Successful.")
    print("New Balance:", bank[account]["bal"])


# -----------------------------
# Withdraw
# -----------------------------
def withdraw(bank, account):
    amount = float(input("Enter Withdraw Amount: "))

    if amount <= 0:
        print("Invalid amount.")
        return

    if amount > bank[account]["bal"]:
        print("Insufficient Balance.")
    else:
        bank[account]["bal"] -= amount
        print("Withdrawal Successful.")
        print("Remaining Balance:", bank[account]["bal"])


# -----------------------------
# Login
# -----------------------------
def login(bank):
    account = input("Enter Account Number: ")
    password = input("Enter Password: ")

    if account not in bank:
        print("Account not found.")
        return

    if bank[account]["password"] != password:
        print("Incorrect Password.")
        return

    print(f"\nWelcome {bank[account]['name']}")

    while True:
        print("\n===== ACCOUNT MENU =====")
        print("1. Balance Inquiry")
        print("2. Deposit")
        print("3. Withdraw")
        print("4. Logout")

        choice = input("Enter Choice: ")

        if choice == "1":
            balance_inquiry(bank, account)

        elif choice == "2":
            deposit(bank, account)

        elif choice == "3":
            withdraw(bank, account)

        elif choice == "4":
            print("Logged Out Successfully.")
            break

        else:
            print("Invalid Choice.")


# -----------------------------
# Main Program
# -----------------------------
def main():
    bank = load_data()

    while True:
        print("\n========== BANK MANAGEMENT ==========")
        print("1. Create New Account")
        print("2. Login")
        print("3. Exit")

        choice = input("Enter Choice: ")

        if choice == "1":
            create_account(bank)

        elif choice == "2":
            login(bank)

        elif choice == "3":
            save_data(bank)
            print("Data Saved Successfully.")
            print("Thank You for Using Bank Management System.")
            break

        else:
            print("Invalid Choice.")


# -----------------------------
# Run Program
# -----------------------------
main()
```

```python
import os
import pandas as pd

# Load existing data
if not os.path.exists("bank1.csv") or os.path.getsize("bank1.csv") == 0:
    df = pd.DataFrame(columns=["A/C # ", "name", "password", "A/C bal"])
else:
    df = pd.read_csv("bank1.csv")


# Store data in dictionary
bank = {}

for i in range(len(df)):
    account = str(df["A/C # "][i])
    bank[account] = {
        "name": df["name"][i],
        "password": str(df["password"][i]),
        "bal": float(df["A/C bal"][i])
    }


# Main Program
while True:

    choice = int(input("\n1. Create New Account\n2. Login\n3. Exit\nEnter Choice: "))


    # Create Account
    if choice == 1:

        account = input("Enter account number: ")

        if account in bank:
            print("Account already exists!")

        else:
            bank[account] = {}

            bank[account]["name"] = input("Enter account holder name: ")
            bank[account]["password"] = input("Enter account password: ")
            bank[account]["bal"] = float(input("Enter opening balance: "))

            print("Account created successfully!")


    # Login
    elif choice == 2:

        account = input("Enter account number: ")
        password = input("Enter password: ")


        if account in bank and password == bank[account]["password"]:

            print("\nLogin Successfully!")

            while True:

                option = int(input(
                    "\n1. Balance Inquiry"
                    "\n2. Deposit"
                    "\n3. Withdraw"
                    "\n4. Logout"
                    "\nEnter Choice: "
                ))


                # Balance
                if option == 1:

                    print("Account Holder:", bank[account]["name"])
                    print("Account Balance:", bank[account]["bal"])


                # Deposit
                elif option == 2:

                    amount = float(input("Enter deposit amount: "))

                    if amount > 0:
                        bank[account]["bal"] += amount
                        print("Updated Balance:", bank[account]["bal"])
                    else:
                        print("Invalid amount")


                # Withdraw
                elif option == 3:

                    amount = float(input("Enter withdrawal amount: "))

                    if amount <= 0:
                        print("Invalid amount")

                    elif amount <= bank[account]["bal"]:

                        bank[account]["bal"] -= amount
                        print("Remaining Balance:", bank[account]["bal"])

                    else:
                        print("Insufficient Balance")


                # Logout
                elif option == 4:

                    print("Logout Successfully!")
                    break


                else:
                    print("Invalid Choice")


        else:
            print("Incorrect Account Number or Password")


    # Exit and Save Data
    elif choice == 3:

        accounts = []
        names = []
        passwords = []
        balances = []


        for account in bank:

            accounts.append(account)
            names.append(bank[account]["name"])
            passwords.append(bank[account]["password"])
            balances.append(bank[account]["bal"])


        df = pd.DataFrame({
            "A/C # ": accounts,
            "name": names,
            "password": passwords,
            "A/C bal": balances
        })


        df.to_csv("bank1.csv", index=False)

        print("Data Saved Successfully!")
        print("Thank you for using Bank System")

        break


    else:
        print("Invalid Choice")
```
