```python
import pandas as pd
que = pd.read_csv("Questions.csv")
randomeQuestions = que.sample(1)
questions = list(dict(randomeQuestions)["Question"])
OptionsAs = list(dict(randomeQuestions)["Option A"])
OptionsBs = list(dict(randomeQuestions)["Option B"])
OptionsCs = list(dict(randomeQuestions)["Option C"])
OptionsDs = list(dict(randomeQuestions)["Option D"])
RightAnswers = list(dict(randomeQuestions)["Right Answer"])
name = input("Enter your name: ").title()
import os
if os.path.exists("result.csv"):
    results = pd.read_csv("result.csv")

    if "Name" in results.columns:
        if name in results["Name"].values:
            print("You have already taken this quiz!")
            exit()
print("Welcome to Python Clas Quiz,Lets start the quiz.")
score = 0
for i in range(len(questions)):
    print(f"\nQuestion {i+1}")
    print(questions[i])

    print("A.", OptionsAs[i])
    print("B.", OptionsBs[i])
    print("C.", optionsCs[i])
    print("D.", optionsDs[i])

    answer = input("Enter your answer (A/B/C/D): ").upper()
    if answer == RightAnswers[i]:
        print("Correct")
        score += 1
    else:
        print("Wrong")
        print("Correct Answer:", RightAnswers[i])
print(f"Your Name is: {name}")
print(f"Your Score is : {score}")
import os

result = pd.DataFrame({
    "Name": [name],
    "Score": [score]
})

file_exists = os.path.isfile("result.csv")

result.to_csv(
    "result.csv",
    mode="a",
    index=False,
    header=not file_exists
)

print("Result saved successfully!")
```
